## Setup

```python
import sys

sys.path.append("../utils")

import matplotlib.pyplot as plt
import pandas as pd
from data_loader import calculate_tau_star, load_all
```

---

## Daten laden

```python
datasets = load_all()

for name, content in datasets.items():
    print(f"{name}:")
    print("Metadata:", content["metadata"])
    if content["data"] is not None:
        print("Data sample:")
        print(content["data"].head())
    print("\n")
```

---

## Beispiel: τ*-Berechnung

```python
# Beispiel für Urban Heat
# Note: Diese Werte sind aus der UTAC-Analyse
beta = 16.3
theta = 10
R = 9.3

tau_star = calculate_tau_star(beta, theta, R)
print(f"τ* (Urban Heat): {tau_star:.2f} Jahre")
```

---

## Visualisierung: β vs τ*

```python
systems = {
    "Urban Heat": (16.3, 10, 9.3),
    "Amazon Moisture": (14.6, 10, 8.8),
    "AMOC": (4.0, 10, 5.0),
    "WAIS": (5.3, 10, 7.5)
}

results = []
for name, (beta, theta, R) in systems.items():
    tau_star = calculate_tau_star(beta, theta, R)
    results.append({"System": name, "β": beta, "τ*": tau_star})

df = pd.DataFrame(results)

plt.figure(figsize=(8,6))
plt.scatter(df["β"], df["τ*"], s=100, c="red")
for i, row in df.iterrows():
    plt.text(row["β"]+0.2, row["τ*"], row["System"])
plt.xlabel("β (Steilheit)")
plt.ylabel("τ* (Jahre)")
plt.title("UTAC – β vs τ* für kritische Systeme")
plt.grid(True)
plt.show()
```

---

## Interpretation

- **Hohe β-Werte (>14)** → sehr kurze τ* (wenig Zeitfenster).  
- **Moderate β-Werte (4–6)** → längere τ*, mehr Handlungsspielraum.  
- **Visualisierung** zeigt, welche Systeme akut gefährdet sind und wo Prävention am dringendsten ist.
